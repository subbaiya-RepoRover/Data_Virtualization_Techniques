```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the games dataset
df_chess = pd.read_csv("/content/games.csv")

# Display the first 5 rows of the dataframe
display(df_chess.head())
```

```python
# Display basic information about the dataset
print(df_chess.info())
```

```python
# Display descriptive statistics for numerical columns
display(df_chess.describe())
```

```python
# Check for missing values
display(df_chess.isnull().sum())
```

### Distribution of Game Turns

```python
plt.figure(figsize=(10, 6))
sns.histplot(df_chess['turns'], bins=50, kde=True)
plt.title('Distribution of Number of Turns per Game')
plt.xlabel('Number of Turns')
plt.ylabel('Frequency')
plt.show()
```

### Relationship between White and Black Player Ratings

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='white_rating', y='black_rating', data=df_chess, alpha=0.5)
plt.title('White Player Rating vs. Black Player Rating')
plt.xlabel('White Player Rating')
plt.ylabel('Black Player Rating')
plt.show()
```

### Distribution of Winners

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='winner', data=df_chess, palette='viridis')
plt.title('Distribution of Game Winners')
plt.xlabel('Winner')
plt.ylabel('Number of Games')
plt.show()
```

### Distribution of Victory Status

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='victory_status', data=df_chess, palette='magma')
plt.title('Distribution of Victory Status')
plt.xlabel('Victory Status')
plt.ylabel('Number of Games')
plt.xticks(rotation=45)
plt.show()
```

### Outlier Detection in Player Ratings (Box Plots)

```python
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
sns.boxplot(y=df_chess['white_rating'], color='skyblue')
plt.title('White Player Rating Distribution')

plt.subplot(1, 2, 2)
sns.boxplot(y=df_chess['black_rating'], color='lightcoral')
plt.title('Black Player Rating Distribution')
plt.tight_layout()
plt.show()
```

### Correlation Heatmap of Numerical Features

```python
numerical_cols = ['turns', 'white_rating', 'black_rating', 'opening_ply']
corr_chess = df_chess[numerical_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr_chess, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```

### Top 10 Most Frequent Opening Moves

```python
top_openings = df_chess['opening_name'].value_counts().head(10)

plt.figure(figsize=(12, 6))
sns.barplot(x=top_openings.index, y=top_openings.values, palette='plasma')
plt.title('Top 10 Most Frequent Chess Opening Moves')
plt.xlabel('Opening Name')
plt.ylabel('Number of Games')
plt.xticks(rotation=75, ha='right')
plt.tight_layout()
plt.show()
```
